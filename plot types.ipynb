```python
import matplotlib.pyplot as plt 
x=[1,2,3,4,5]
y=[2,4,6,8,15]
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("Line Plot Example")
plt.plot(x,y)
plt.savefig("sample.png")
plt.show()
```

```python
import matplotlib.pyplot as plt 
x=[101,102,103,104,105]
y=[66,87,96,88,50]
plt.xlabel("Student rollno")
plt.ylabel("MATHS Marks")
plt.xlim(101,105)
plt.ylim(10,100)
plt.title("Student Performance")
plt.plot(x,y)
plt.plot(x,y,color='brown',linestyle='dashdot',linewidth=3,marker='v',markerfacecolor='grey',markersize=12)
plt.savefig("mmarks1.png")
plt.show()
```

```python
import matplotlib.pyplot as plt 
x=[1,2,3,4,5]
y=[2,4,6,8,10]
x1=[1,2,3,4]
y1=[4,1,3,6]
plt.xlabel("Student rollno")
plt.ylabel("MATHS Marks")
plt.xlim(1,10)
plt.ylim(2,12)
plt.title("Student Performance")
plt.plot(x,y)
plt.plot(x,y,label="Line-1",color='brown',linestyle='dashdot',linewidth=3,marker='v',markerfacecolor='grey',markersize=12)
plt.plot(x1,y1,label="Line-2",color='red',linestyle='dotted')
plt.legend()
plt.savefig("mmarks1.png")
plt.show() 
```

```python
import matplotlib.pyplot as plt 
students=["A","B","C","D","E"]
marks=[65,76,44,77,55]
plt.bar(students,marks,color="blue",width=0.4,edgecolor="red")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks in Semester")
plt.savefig("bar1.png")
plt.show() 
```
