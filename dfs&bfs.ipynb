```python
graph={
    0:[1,2],
    1:[0,3,4],
    2:[0,5],
    3:[1],
    4:[1,5],
    5:[2,4]
}
#dfs
def dfs(graph,start,visited=None):
   if visited is None:
      visited=set()
      
   visited.add(start)
   print(start,end="")

   for neighbour in graph[start]:
      if neighbour not in visited:
         dfs(graph,neighbour,visited)

print("DFS Traversal:")
dfs(graph,0)

def bfs(graph,start):
      visited=set()
      queue=[start]
      visited.add(start)
      while queue:
          vertex=queue.pop(0)
          print(vertex,end="")
          for neighbour in graph[start]:
              if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighnour)

print("BFS Traversal:")
dfs(graph,0)
  
    
```
